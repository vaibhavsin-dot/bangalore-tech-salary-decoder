```python
def parse_ctc(value):
    """
    Parses Current_CTC / Previous_CTC strings into a float in LPA (Lakhs Per Annum).
    Formats seen: '15.5 LPA', '15.5', '15,50,000', '₹15.5 LPA'
    Strategy: strip currency symbols/text, remove commas, convert to float,
    then check magnitude — if it looks like a raw rupee figure (large number),
    divide by 100000 to convert to LPA.
    """
    if pd.isna(value):
        return np.nan  # preserve NaN for freshers (Previous_CTC)

    s = str(value)
    s = s.replace('₹', '').replace('LPA', '').replace(',', '').strip()

    try:
        num = float(s)
    except ValueError:
        return np.nan  # unparseable value, coerce to NaN

    # If the number is large, it's a raw rupee figure, not LPA
    if num > 1000:
        num = num / 100000

    return num

# Apply to both CTC columns
df['current_ctc_lpa'] = df['Current_CTC'].apply(parse_ctc)
df['previous_ctc_lpa'] = df['Previous_CTC'].apply(parse_ctc)

print(df[['Current_CTC', 'current_ctc_lpa', 'Previous_CTC', 'previous_ctc_lpa']].head(10))
```

```python
# Q3.1: CTC distribution by role — median, mean, min, max, sorted by median descending
role_ctc_stats = df.groupby('role_clean')['current_ctc_lpa'].agg(['median', 'mean', 'min', 'max'])
role_ctc_stats = role_ctc_stats.sort_values('median', ascending=False)
print(role_ctc_stats)
```

```python
# Q3.2: Experience curve for SDE Backend — bin years_exp into 4 bands, compute median CTC per band
sde_backend = df[df['role_clean'] == 'SDE Backend'].copy()

# Define bin edges: 0-1, 2-3, 4-5, 6+
bins = [-1, 1, 3, 5, 100]
labels = ['0-1', '2-3', '4-5', '6+']
sde_backend['exp_band'] = pd.cut(sde_backend['years_exp'], bins=bins, labels=labels)

exp_median = sde_backend.groupby('exp_band', observed=True)['current_ctc_lpa'].median()
print(exp_median)

# Compute % growth from one band to the next
growth = exp_median.pct_change() * 100
print(growth)
```

```python
# Q3.3: Skill premium for SDEs — compare median CTC with vs without each skill
sde_roles = ['SDE Backend', 'SDE Frontend', 'SDE Full-Stack']
sdes = df[df['role_clean'].isin(sde_roles)].copy()

skills_to_check = ['AWS', 'ML', 'System Design', 'Kubernetes']

for skill in skills_to_check:
    # .str.contains checks if the skill appears anywhere in the comma-separated Skills string
    # na=False treats NaN skill entries as "does not have the skill"
    has_skill = sdes['Skills'].str.contains(skill, case=False, na=False)
    
    median_with = sdes.loc[has_skill, 'current_ctc_lpa'].median()
    median_without = sdes.loc[~has_skill, 'current_ctc_lpa'].median()
    premium_pct = ((median_with - median_without) / median_without) * 100
    
    print(f"{skill:<15} With: {median_with:.1f} LPA   Without: {median_without:.1f} LPA   Premium: {premium_pct:+.1f}%")
```

```python
# Q3.4: Company-type premium for SDE Backend — median CTC by company type
target_role = df[df['role_clean'] == 'SDE Backend']

company_premium = target_role.groupby('company_type_clean')['current_ctc_lpa'].median()
company_premium = company_premium.sort_values(ascending=False)
print(company_premium)

# % premium of Unicorn over MNC, same role
unicorn_median = company_premium['Unicorn']
mnc_median = company_premium['MNC']
premium_pct = ((unicorn_median - mnc_median) / mnc_median) * 100
print(f"Unicorn premium over MNC: {premium_pct:+.1f}%")
```

```python
# Q3.5 Step 1: create exp_band for the full dataset (needed to group by role + company_type + exp_band)
bins = [-1, 1, 3, 5, 100]
labels = ['0-1', '2-3', '4-5', '6+']
df['exp_band'] = pd.cut(df['years_exp'], bins=bins, labels=labels)

df['group_median'] = df.groupby(
    ['role_clean', 'company_type_clean', 'exp_band'], observed=True
)['current_ctc_lpa'].transform('median')


df['group_size'] = df.groupby(
    ['role_clean', 'company_type_clean', 'exp_band'], observed=True
)['current_ctc_lpa'].transform('size')
```

```python
# Step 3: gap = individual's CTC minus their group's median. Negative gap = underpaid relative to peers.
df['gap'] = df['current_ctc_lpa'] - df['group_median']

# Only trust groups with at least 10 members (per brief instructions)
reliable = df[df['group_size'] >= 10]

underpaid = reliable.sort_values('gap').head(10)
print(underpaid[['Employee_ID',
                 'role_clean',
                 'company_type_clean',
                 'years_exp',
                 'current_ctc_lpa',
                 'group_median',
                 'gap']])
```

## Key Insights

1. Product Managers earn the highest median CTC at 31.3 LPA, 1.85x the median of Data Analysts, 
   the lowest-paid role at 16.9 LPA. Students choosing between PM-track and analyst-track roles 
   should factor in this near-2x gap when negotiating offers or planning a career pivot.

2. SDE Backend CTC jumps +72% between the 0-1 and 2-3 year experience bands (11.6 LPA to 
   20.0 LPA) — the steepest single jump in the career curve. Students should prioritize crossing 
   the 2-year mark before switching jobs, since that's where the market rewards experience most.

3. System Design carries the highest premium among SDEs at +20.5% (25.3 LPA with vs 21.0 LPA 
   without) — notably higher than Kubernetes (+7.9%) or ML (+12.4%), and AWS actually shows a 
   slight negative premium (-4.3%), suggesting it's become table-stakes rather than a 
   differentiator. Students prepping for SDE interviews should prioritize System Design depth 
   over checkbox cloud certifications, since it shows the clearest measurable payoff.
